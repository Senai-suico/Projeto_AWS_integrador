```python
df_avaliacoes["review_creation_date"] = pd.to_datetime(df_avaliacoes["review_creation_date"])
df_avaliacoes["review_answer_timestamp"] = pd.to_datetime(df_avaliacoes["review_answer_timestamp"])
df_avaliacoes['review_comment_title'] = df_avaliacoes['review_comment_title'].fillna(" ")
df_avaliacoes['review_comment_message'] = df_avaliacoes['review_comment_message'].fillna(" ")
df_avaliacoes['review_comment_message'] = df_avaliacoes['review_comment_title'].astype(str) + ' ' + df_avaliacoes['review_comment_message'].astype(str)
def remover_acentos(texto):
    if not isinstance(texto, str):
        return texto
    return ''.join(c for c in unicodedata.normalize('NFD', texto)
                  if unicodedata.category(c) != 'Mn')


df_avaliacoes['review_comment_message'] = (
    df_avaliacoes['review_comment_message']
    .fillna('')
    .str.lower()
    .apply(remover_acentos)
)
nulo = df_avaliacoes[df_avaliacoes["review_comment_title"].isna() & df_avaliacoes["review_comment_message"].isna()].shape[0]
df_avaliacoes
```
